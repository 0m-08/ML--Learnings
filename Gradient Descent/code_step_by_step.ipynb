```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
X,y=make_regression(n_samples=4, n_features=1,n_informative=1, noise=10, random_state=18)
import matplotlib.pyplot as plt
plt.scatter(X, y)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X,y)
print(model.coef_,model.intercept_)
```

```python
plt.scatter(X,y)
plt.plot(X,model.predict(X),color='red')
```

```python
y_pred = ((78.35 * X) + 100)
     

plt.scatter(X,y)
plt.plot(X,model.predict(X),color='red',label='OLS')
plt.plot(X,y_pred,color='#00a65a',label='b = 0')
plt.legend()
plt.show()
```

Gradient Descent assuming slope is constant m = 78.35 b = 0

```python
b = -100
m = 78.35
lr = 0.01

epochs = 100

for i in range(epochs):
  loss_slope = -2 * np.sum(y - m*X.ravel() - b)
  b = b - (lr * loss_slope)

  y_pred = m * X + b

  plt.plot(X,y_pred)

plt.scatter(X,y)
```
