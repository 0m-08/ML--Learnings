# Ingesting SQL Data into Pandas 🗄️

In real-world data science workflows, data is rarely stored exclusively in CSV files. Most structured business data resides in relational databases (RDBs) like MySQL, PostgreSQL, or SQLite.

This notebook demonstrates how to connect to a relational database using Python and load SQL query results directly into a Pandas DataFrame using `pd.read_sql_query()`.

### 1. Install Database Connector Driver ⚙️
To interact with a database, Python requires a driver compatible with the target RDBMS. Since we are connecting to a **MySQL** database, we install `mysql-connector-python`.

```python
!pip install mysql.connector
```

### 2. Import Libraries 📥
Next, we import our connection driver and `pandas`.

```python
import mysql.connector
import pandas as pd
```

### 3. Establish Connection & Query Database 🔗
We connect to our database by specifying the database credentials. After establishing a connection, we can write a SQL query and use `pd.read_sql_query()` to automatically compile the SQL output into a standard Pandas DataFrame.

```python
# Connect to the local MySQL server
onn = mysql.connector.connect(
    host='localhost',
    user='root',
    password='',
    database='world'
)

# Run the SQL query and load the results into a DataFrame
df = pd.read_sql_query("SELECT * FROM city", onn)
df.head()
```
