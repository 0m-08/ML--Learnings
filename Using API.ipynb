# Fetching Data from APIs 🌐

Application Programming Interfaces (APIs) are widely used to request real-time, dynamic data from external web services. Rather than downloading static files, we programmatically query an API server to fetch information on-demand.

This notebook covers:

*   Importing the Python `requests` library
*   Setting up API endpoints, headers, and authentication parameters
*   Sending standard `GET` requests and parsing the JSON response payload into a structured format

### 1. Import Core Libraries 📥
We import `requests` to handle HTTP transactions and `pandas` to structure data.

```python
import requests
import pandas as pd
```

### 2. Requesting Data from RapidAPI 🚀
Here we execute a structured `GET` request using headers for authentication. In this example, we query the **Streaming Availability API** to fetch metadata for a movie using its IMDb ID.

```python
import requests

# 1. Define endpoint parameters (IMDb ID and media type)
show_type = "movie"
show_id = "tt0111161"  # IMDb ID for The Shawshank Redemption

url = f"https://streaming-availability.p.rapidapi.com/shows/{show_type}/{show_id}"

# 2. Specify authorization headers required by RapidAPI
headers = {
    "x-rapidapi-key": "1cf2ce1a40mshe37b0d292184931p138544jsn97cc08c47459",
    "x-rapidapi-host": "streaming-availability.p.rapidapi.com"
}

# 3. Execute the GET request securely
try:
    response = requests.get(url, headers=headers)
    response.raise_for_status()  # Check for HTTP errors
    
    # 4. Parse the JSON response
    data = response.json()
    print("API Request Successful!")
    print(f"Title: {data.get('title', 'N/A')}")
except Exception as e:
    print(f"An error occurred: {e}")
```
